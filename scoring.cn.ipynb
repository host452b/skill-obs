## 📝 长期维护 / Long-term maintenance

每次重新评分都会向 `build_scoring_notebook.py` 的 `EVALUATIONS` **追加**一个新 dict。一个快照 dict 包含：

- `eval_date` — ISO-8601 UTC 时间戳
- `version` — 人类可读标签
- `submodule_shas` — 评分时刻的 `{路径 → SHA}`（从 `git submodule status` 复制）
- `raw_metrics` — `{code → {指标 → 值}}`（gh API + 本地 `find`）
- `scores` — `{code → {维度id → 1-10 整数}}`（基于排名）

**切勿修改历史。** 旧快照作为审计轨迹留存。当存在 ≥2 个快照时，Δ-diff 部分会自动渲染。

**新增 cohort 成员**：在 `REPOS` 列表注册，并在每个快照里为它补分（旧快照未评测则用 `null`），再重跑构建脚本。
